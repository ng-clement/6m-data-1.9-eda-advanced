```python
import pandas as pd
import numpy as np
from datetime import datetime

print("Create a dummy dataframe (or load one) with an index of daily dates for the year 2023.\n")
df_dates = pd.date_range(start="2023-01-01", end="2023-12-31", freq="D")
print(df_dates)
print()

print("Generate random \"Sales\" data for each day.\n")
sales_data = np.random.randint(100, 1000, size=len(df_dates))  # Random sales between 100 and 1000
df_day_sale = pd.Series(sales_data, index=df_dates)
print(df_day_sale.head())
print()

print("Resample this daily data to find the Total (Sum) Sales per Month.\n")
monthly_sales = df_day_sale.resample('M').sum()
print(monthly_sales)
```

**Task 2: The "Wide" Report**

1. Create a small dataframe with columns: \['Date', 'Region', 'Sales'\]. Populate it with data for 3 regions (North, South, East) over 5 different dates.  
2. Use **pivot()** to reshape this data so that:  
   * Date is the Index.  
   * Region names are the Columns (North, South, East).  
   * Sales are the Values.

```python
print("Create a small dataframe with columns: \['Date', 'Region', 'Sales'\]. Populate it with data for 3 regions (North, South, East) over 5 different dates.\n")
data = {
    'Date': pd.date_range("2023-01-01", periods=5).tolist() * 3,
    'Region': ['North'] * 5 + ['South'] * 5 + ['East'] * 5,
    'Sales': np.random.randint(100, 1000, size=15)
} 
df = pd.DataFrame(data)
print(df)
print()

print("Use **pivot()** to reshape this data so that: ")
print("   * Date is the Index.  ")
print("   * Region names are the Columns (North, South, East).")  
print("   * Sales are the Values.\n")
pivot_df = df.pivot(index='Date', columns='Region', values='Sales')
print(pivot_df)
```

   * **Rows:** Day of the Week.

```python
df_tips = pd.read_csv("../data/tips.csv")
df_tips.head(10)
```

```python
summary_table = df_tips.pivot_table(
    index='day', 
    columns='time', 
    values='total_bill', 
    aggfunc='mean',
    margins=True,
    margins_name='All'
)
print(summary_table)
```
